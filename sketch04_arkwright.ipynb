```python
# geopandas does not need to be gpd but this code imports geopandas which is a libary for dealing with spatial data
# Same goes for pandas reading as pd
# geopandas is better suited for spatial data and pandas is better suited for data in tables like a csv file
# Therefore I am commenting out importing geopandas and instead am importing pandas


### import geopandas as gpd ###
import pandas as pd


# The code below imports ohio_precip_2025.csv from the data in Dr. Bittermans repository
# I could load the data into my own and work from there but that requires downloading that is not really necessary
precip = pd.read_csv(f"{DATA_LOCATION}/part1/ohio_precip_2025.csv")

# For the code below the last code is automatically printed on the output section (hence .describe showing up) to get the table showing the column names (using .head) to also show I needed to wrap it in a print
# Length tells me how many stations there are
# .head lists the first 5 of those stations in the list as well as their column names
# .describe lists the statistical info about the data in the file

# Using display instead of print allows me to make tables I can actually hover over rather than smaller print that is just an output
print(f"There are {len(precip)} precipitation stations in Ohio.")
precip.head()
display (precip.head())
precip["precip_mm_2025"].describe()
display (precip["precip_mm_2025"].describe())


# Using the parameters feature in Google Colab I was able to figure out the parameters needed to sort the 5 stations with the 5 highest precip values
# Setting ascedning to False tells Python to sort through the list with the parameter of "precip_mm_2025" and sort it from largest to smallest
# Setting ascending to True (the default) tells Python to sort the list items from least to greatest
# Using .head(5) give the top 5 stations

# I put both name and precip_mm_2025 into the double brackets to get two columns as the output/print
wettest = precip.sort_values("precip_mm_2025", ascending=False).head(5)
display (wettest[["name", "precip_mm_2025"]])

max_precip = precip["precip_mm_2025"].max()
mean_precip = precip["precip_mm_2025"].mean()

difference = max_precip - mean_precip

print(f"The station precip mean is {mean_precip:.2f}mm with the max precip recorded meaning {max_precip}mm. Therefore the station with the max precip recorded was {difference:.2f}mm over the mean.")
```

The station with the max precip readings was 486.84mm over the mean which is not really suprising. The stations that we used are from 2025 and cover all of those in Ohio, therefore, one area of Ohio may have gotten more precipitation than another area that is typically more dry. The weather is a weird thing and sometimes some areas get substantially more rain, weather events, etc. than other areas.
